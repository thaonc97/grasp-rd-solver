# This notebook is for doing experiments with the code

```python
import logging
import numpy as np

from config import DEFAULT_CONFIG, Config
from examples import loader
from heuristics import GRASPSolver, RandomAssignmentSolver
from logging_utils import setup_logging
from models import RegisterInstance
from scripts import plot_history

import matplotlib.pyplot as plt
```

```python
atoms_example = 50 # number of atoms
layout_size_example = 100 # number of trapping sites
instance_location = f"examples/instances/instance_{atoms_example}_atoms_{layout_size_example}_sites"

max_iterations = 50

SEED = 52
experiment_config = Config(seed=SEED)
setup_logging(logging.INFO)
```

```python
q,layouts = loader.load_example_instance(instance_location)

#solve the layouts using the GRASP
grasp_solve_results = GRASPSolver.solve_rd(layouts,q, max_iterations=max_iterations, config=experiment_config)
#solve the layouts using the RandomAssignment
random_solve_results = RandomAssignmentSolver.solve_layouts(layouts,q, max_iterations=max_iterations, config=experiment_config)
```

```python
#Plot the history of the best_results

best_history = grasp_solve_results.best_result.history
plt.plot(best_history)
plt.xlabel("Iteration")
plt.ylabel("Objective Value")
plt.title("History of Best Result")
plt.show()
```
